```python
from google.colab import userdata

import os
os.environ['AWS_ACCESS_KEY_ID'] = ''
os.environ['AWS_SECRET_ACCESS_KEY'] = ''
```

```python
aws_access_key_id = os.environ.get('AWS_ACCESS_KEY_ID')
aws_secret_access_key = os.environ.get('AWS_SECRET_ACCESS_KEY')

# Initialize S3 client using the retrieved credentials
s3_client = boto3.client('s3', region_name='us-east-1',
                         aws_access_key_id=aws_access_key_id,
                         aws_secret_access_key=aws_secret_access_key)

# Define S3 bucket name and local directory
bucket_name = 'deepseemodel'
local_dir = 'DeepSeek-R1-Distill-Llama-8B'

# Upload files from local directory to S3
for root, dirs, files in os.walk(local_dir):
    for file in files:
        local_path = os.path.join(root, file)
        s3_key = os.path.relpath(local_path, local_dir)
        s3_client.upload_file(local_path, bucket_name, s3_key)
        print(f'Uploaded {file} to s3://{bucket_name}/{s3_key}')
```

```python
import boto3
import json

model_arn = ""
prompt = "What is a noun?"

# Ensure the AWS credentials are properly passed
client = boto3.client(service_name='bedrock-runtime',
                      region_name='us-east-1',
                      aws_access_key_id=aws_access_key_id,
                      aws_secret_access_key=aws_secret_access_key)

body = json.dumps({
    'prompt': prompt,
    'max_tokens_to_sample': 4000
})

response = client.invoke_model_with_response_stream(
    modelId=model_arn,
    body=body
)

stream = response.get('body')
if stream:
    for event in stream:
        chunk = event.get('chunk')
        if chunk:
            print(json.loads(chunk.get('bytes').decode())['generation'], end='')
```
